```python
product_info=pd.DataFrame({
    "Product Name":[
        "Laptop",
        "Mouse",
        "Printer",
        "Tablet",
        "Monitor"
    ],
    
    "Supplier":[
        "Dell",
        "Logitech",
        "HP",
        "Samsung",
        "LG"
    ],
    
    
    "Discount":[
        10,
        5,
        15,
        8,
        12
    ]
})

print("13.PRODUCT INFORMATION DATAFRAME")
print("-"*50)
display(product_info)
```

```python
merged_df=pd.merge(
    df,
    product_info,
    on="Product Name",
    how="left"
)

print("14.MAERGED DATAFRAME")
print("-"*50)
display(merged_df)
```

```python
region_info=pd.DataFrame({
    "Region_Manager":[
        "Manager A",
        "Manager B",
        "Manager C",
        "Manager D"
        
    ]    
    
},
    
    index=[
        "North",
        "South",
        "East",
        "West"
    ]
)

print("15.REGION INFORMATION")
print("-"*50)
display(region_info)
```

```python
df_region = df.set_index("Region")

joined_df = df_region.join(region_info)

print("16.JOINED DATAFRAME")
print("-"*50)
display(joined_df)
```

```python
region_sales = df.groupby(
    "Region"
)["Sales"].sum()

print("16.TOTAL SALES BY REGION")
print("-"*50)
display(region_sales)
```

```python
pivot_table=pd.pivot_table(
    df,
    values="Sales",
    index="Category",
    columns="Region",
    aggfunc="sum",
    fill_value=0
)

print("18.PIVOTE TABLE")
print("-"*50)
display(pivot_table)
```
